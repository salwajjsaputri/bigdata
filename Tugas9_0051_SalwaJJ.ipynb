```python
# Tugas Mandiri: Katalog Menu Kopi Nusantara dengan MongoDB dan PySpark
```

```python
# A. Membangun Katalog (Create)
# Instruksi: Buat database kopi_nusantara, koleksi menu, dan minimal 12 dokumen dengan field yang sesuai kategorinya.
```

```python
from pymongo import MongoClient

# 1. Koneksi ke MongoDB
client = MongoClient("mongodb://localhost:27017/")
client.admin.command("ping")
print("Koneksi MongoDB berhasil!")

# 2. Membuat atau memilih database dan collection
db = client["kopi_nusantara"]
koleksi_menu = db["menu"]

# 3. Menghapus data lama agar praktikum dimulai dari kondisi bersih
koleksi_menu.delete_many({})

# 4. Menyiapkan minimal 12 data menu
data_menu = [
    # Kategori Kopi
    {"nama_menu": "Espresso", "harga": 15000, "kategori": "Kopi",
     "tingkat_kepahitan": 5},
    {"nama_menu": "Americano", "harga": 18000, "kategori": "Kopi",
     "tingkat_kepahitan": 4},
    {"nama_menu": "Cappuccino", "harga": 25000, "kategori": "Kopi",
     "tingkat_kepahitan": 3},
    {"nama_menu": "Kopi Susu Gula Aren", "harga": 22000, "kategori": "Kopi",
     "tingkat_kepahitan": 2},

    # Kategori Makanan
    {"nama_menu": "Nasi Goreng", "harga": 28000, "kategori": "Makanan",
     "pedas_level": 3},
    {"nama_menu": "Mie Goreng", "harga": 20000, "kategori": "Makanan",
     "pedas_level": 2},
    {"nama_menu": "Roti Bakar", "harga": 15000, "kategori": "Makanan",
     "pedas_level": 0},
    {"nama_menu": "Kentang Goreng", "harga": 18000, "kategori": "Makanan",
     "pedas_level": 0},

    # Kategori Minuman nonkopi
    {"nama_menu": "Teh Manis", "harga": 10000, "kategori": "Minuman",
     "stok": 30, "rekomendasi_barista": "Cocok diminum dingin"},
    {"nama_menu": "Matcha Latte", "harga": 25000, "kategori": "Minuman",
     "stok": 20, "rekomendasi_barista": "Favorit pelanggan"},
    {"nama_menu": "Cokelat Dingin", "harga": 22000, "kategori": "Minuman",
     "stok": 25, "rekomendasi_barista": "Cocok untuk pencinta cokelat"},
    {"nama_menu": "Lemon Tea", "harga": 15000, "kategori": "Minuman",
     "stok": 35, "rekomendasi_barista": "Rasa segar"}
]

# 5. Memasukkan seluruh data ke MongoDB
hasil = koleksi_menu.insert_many(data_menu)

print("Jumlah data yang dimasukkan:", len(hasil.inserted_ids))
print("Jumlah dokumen dalam koleksi:", koleksi_menu.count_documents({}))

# Menampilkan semua data
for menu in koleksi_menu.find({}, {"_id": 0}):
    print(menu)
```

```python
# B. Kueri Bertingkat
```

```python
# 1. Menu kategori Kopi dengan tingkat kepahitan di atas 3
print("1. Menu kopi dengan tingkat kepahitan di atas 3")

hasil_kopi = koleksi_menu.find({
    "kategori": "Kopi",
    "tingkat_kepahitan": {"$gt": 3}
}, {"_id": 0})

for menu in hasil_kopi:
    print(menu)


# 2. Menu dengan harga antara 15.000 dan 30.000
print("\n2. Menu dengan harga 15.000 sampai 30.000")

hasil_harga = koleksi_menu.find({
    "harga": {"$gte": 15000, "$lte": 30000}
}, {"_id": 0})

for menu in hasil_harga:
    print(menu)


# 3. Menu yang tidak memiliki field pedas_level
print("\n3. Menu tanpa field pedas_level")

hasil_pedas = koleksi_menu.find({
    "pedas_level": {"$exists": False}
}, {"_id": 0})

for menu in hasil_pedas:
    print(menu)
```

```python
# C. Pembaruan Data (Update)
```

```python
# 1. Menambahkan status Tersedia ke semua dokumen
hasil_update = koleksi_menu.update_many(
    {},
    {"$set": {"status": "Tersedia"}}
)

print("Jumlah dokumen yang diperbarui:",
      hasil_update.modified_count)


# 2. Mengubah status Espresso menjadi Habis
hasil_espresso = koleksi_menu.update_one(
    {"nama_menu": "Espresso"},
    {"$set": {"status": "Habis"}}
)

print("Espresso ditemukan:", hasil_espresso.matched_count)


# 3. Mengubah status Nasi Goreng menjadi Habis
hasil_nasi = koleksi_menu.update_one(
    {"nama_menu": "Nasi Goreng"},
    {"$set": {"status": "Habis"}}
)

print("Nasi Goreng ditemukan:", hasil_nasi.matched_count)


# 4. Menampilkan hasil pembaruan
print("\nStatus menu setelah diperbarui:")

for menu in koleksi_menu.find(
    {},
    {"_id": 0, "nama_menu": 1, "status": 1}
):
    print(menu)
```

```python
# D. Integrasi MongoDB ke Spark DataFrame
```

```python
import pandas as pd

from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType
)
from pyspark.sql.functions import avg, count


# 1. Membuat SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")


# 2. Mengambil data dari MongoDB
data_mongo = list(
    koleksi_menu.find({}, {"_id": 0})
)

print("Jumlah data dari MongoDB:", len(data_mongo))


# 3. Menentukan kolom yang akan digunakan
kolom = [
    "nama_menu",
    "harga",
    "kategori",
    "tingkat_kepahitan",
    "pedas_level",
    "stok",
    "rekomendasi_barista",
    "status"
]


# 4. Menyamakan field setiap dokumen
data_normal = [
    {k: dokumen.get(k, None) for k in kolom}
    for dokumen in data_mongo
]


# 5. Mengubah data menjadi Pandas DataFrame
df_pandas = pd.DataFrame(data_normal, columns=kolom)

print("\nPandas DataFrame:")
print(df_pandas.head())


# 6. Menentukan skema Spark secara eksplisit
skema = StructType([
    StructField("nama_menu", StringType(), True),
    StructField("harga", IntegerType(), True),
    StructField("kategori", StringType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("stok", IntegerType(), True),
    StructField("rekomendasi_barista", StringType(), True),
    StructField("status", StringType(), True)
])


# 7. Membersihkan nilai kosong dan menyesuaikan tipe data
kolom_integer = [
    "harga",
    "tingkat_kepahitan",
    "pedas_level",
    "stok"
]

records = []

for _, baris in df_pandas.iterrows():
    dokumen = {}

    for k in kolom:
        nilai = baris[k]

        # Mengubah NaN atau nilai kosong menjadi None
        if pd.isna(nilai):
            dokumen[k] = None

        # Memastikan kolom angka menggunakan integer Python
        elif k in kolom_integer:
            dokumen[k] = int(nilai)

        # Memastikan kolom teks menggunakan string Python
        else:
            dokumen[k] = str(nilai)

    records.append(dokumen)


# 8. Membuat Spark DataFrame
df_spark = spark.createDataFrame(
    records,
    schema=skema
)


# 9. Menampilkan skema dan isi data
print("\nSkema Spark DataFrame:")
df_spark.printSchema()

print("\nIsi Spark DataFrame:")
df_spark.show(truncate=False)


# 10. Menghitung rata-rata harga dan jumlah menu per kategori
print("\nRata-rata harga dan jumlah menu per kategori:")

df_spark.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show(truncate=False)
```
